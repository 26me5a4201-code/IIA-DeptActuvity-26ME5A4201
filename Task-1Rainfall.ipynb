```python
import matplotlib.pyplot as plt

# Rainfall data in mm
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
rainfall = [20, 30, 45, 60, 100, 150]

# Display rainfall data
print("Monthly Rainfall Analysis")

for i in range(len(months)):
    print(months[i], ":", rainfall[i], "mm")

# Calculate total and average rainfall
print("Total Rainfall:", sum(rainfall), "mm")
print("Average Rainfall:", sum(rainfall) / len(rainfall), "mm")

# Plot graph
plt.bar(months, rainfall)
plt.xlabel("Months")
plt.ylabel("Rainfall (mm)")
plt.title("Monthly Rainfall Analysis")
plt.show()
```

```python
months = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
rainfall_mm = [12, 8, 20, 45, 90, 140, 180, 160, 210, 250, 130, 40]

total = sum(rainfall_mm)
average = total / len(rainfall_mm)
wettest = months[rainfall_mm.index(max(rainfall_mm))]
driest = months[rainfall_mm.index(min(rainfall_mm))]
above_avg = [m for m, r in zip(months, rainfall_mm) if r > average]

print('Annual rainfall (mm):', total)
print('Monthly average (mm):', round(average, 1))
print('Wettest month:', wettest, '| Driest month:', driest)
print('Months above average:', above_avg)
```
